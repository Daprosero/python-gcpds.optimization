```python
from gekko import GEKKO

# Crear el modelo GEKKO
m = GEKKO(remote=True)

# Definir las variables
x = m.Var(value=1, lb=0, ub=10)
y = m.Var(value=1, lb=0, ub=10)

# Definir las ecuaciones y el objetivo
m.Equation(x + y == 10)
m.Minimize((x-3)*2 + (y-2)*2)

# Configurar el solucionador IPOPT y las opciones
m.options.SOLVER = 3  # Usar IPOPT


# Resolver el modelo
m.solve(disp=True)

# Mostrar resultados
print(f'x: {x.value[0]}')
print(f'y: {y.value[0]}')
```
